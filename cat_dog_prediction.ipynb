```python
import tensorflow

from keras.layers import Dense, Conv2D, MaxPooling2D, Flatten
from keras.models import Sequential
```

# Making CNN Model

```python
# Making CNN model
cnn = Sequential()

# Only one time input_shape is given
cnn.add(Conv2D(32, (3, 3), input_shape=(64, 64, 3), activation='relu'))
cnn.add(MaxPooling2D(pool_size=(2, 2)))

cnn.add(Conv2D(16, (3, 3), activation='relu'))
cnn.add(MaxPooling2D(pool_size=(2, 2)))

cnn.add(Flatten())
```

# Making ANN Model

```python
# Making ANN model
cnn.add(Dense(64, activation='relu'))
cnn.add(Dense(32, activation='relu'))
cnn.add(Dense(16, activation='relu'))
cnn.add(Dense(8, activation='relu'))
cnn.add(Dense(4, activation='relu'))

# output
cnn.add(Dense(1, activation='sigmoid'))
```

# Compiling Model

```python
cnn.compile(loss='binary_crossentropy',optimizer="adam")
```

# Training the Model

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Go to https://faroit.com/keras-docs/1.2.0/preprocessing/image/
# And copy Example of using .flow_from_directory(directory) code:
```

```python
train_datagen = ImageDataGenerator(
        rescale=1./255,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True)

test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
        r"C:\Users\amira\OneDrive\Documents\DEEP_LEARNING\Tutorial (WsCube Tech)\6. CNN (Convolutional Neural Network)\train",
        target_size=(64, 64),
        batch_size=32,
        class_mode='binary')

test_generator = test_datagen.flow_from_directory(
        r"C:\Users\amira\OneDrive\Documents\DEEP_LEARNING\Tutorial (WsCube Tech)\6. CNN (Convolutional Neural Network)\test",
        target_size=(64, 64),
        batch_size=32,
        class_mode='binary')


# more steps_per_epoch means more number of images are going to be trained
# we are taking steps_per_epoch and epochs less in number because traing big model will take a lot of time
cnn.fit(train_generator, steps_per_epoch=200, epochs=30, validation_data=test_generator)
```
